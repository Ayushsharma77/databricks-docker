```python
import json
import random
import time
from kafka import KafkaProducer

# Initialize the Kafka producer using your working network setup
producer = KafkaProducer(
    bootstrap_servers=['kafka:9092'],
    value_serializer=lambda v: json.dumps(v).encode('utf-8')
)

user_ids = [101, 102, 103, 104, 105]
print("Starting Transaction Producer...")

# Simulate 100 transactions
for i in range(100):
    tx_data = {
        "tx_id": i,
        "userId": random.choice(user_ids),
        "amount": round(random.uniform(10.0, 15000.0), 2),  # >10000 triggers fraud
        "timestamp": time.time()
    }
    
    producer.send('fraud-detection', value=tx_data)
    print(f"Sent: {tx_data}")
    time.sleep(2)

producer.flush()
print("Finished sending all transactions.")
```
